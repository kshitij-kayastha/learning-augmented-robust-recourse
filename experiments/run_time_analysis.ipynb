```python
import numpy as np
import pandas as pd
import os
import plotly.express as px
from sklearn.model_selection import ParameterGrid

from src.data import *
from src.utils import *
from src.model import *
from src.recourse import *
```

```python
algorithms = ['Alg1', 'L1PSD', 'ROARLInf', 'ROARL1']
dataset_lamb = {"lr_german": [0.3,0.04, 0.004, 0.1, 0.01],
                "nn_german": [3.0, 0.7, 0.3, 0.1, 0.05, 0.01],
                "lr_sba": [0.01, 0.1, 0.7, 1.4, 2.1],
                "nn_sba": [0.01, 0.1, 0.7, 2.1, 3.5]}
# pair_one = ["lr_german", "nn_german"]
# pair_two = ["lr_sba", "nn_sba"]
pair_one = ["lr_german"]
pair_two = ["lr_sba"]
```

# German

```python
first_time = True

for alg in algorithms:
    data = None
    for val in pair_one:
        for lamb in dataset_lamb[val]:
            for seed in range(5):
                file_path = f"../results/recourse_time/{val}_{alg}_{lamb}_0.5_{seed}_time.pkl"
                ret = pd.read_pickle(file_path)
                if first_time:
                    data = ret.copy(deep=True)               
                    first_time = False
                else:
                    data = pd.concat((data, ret), ignore_index=True)

    time_mean = data["e_time"].mean()
    print(f"{alg} : {time_mean.round(4)} s")
```

```python
first_time = True

for alg in algorithms:
    data = None
    for val in pair_two:
        for lamb in dataset_lamb[val]:
            for seed in range(5):
                file_path = f"../results/recourse_time/{val}_{alg}_{lamb}_0.5_{seed}_time.pkl"
                ret = pd.read_pickle(file_path)
                if first_time:
                    data = ret.copy(deep=True)               
                    first_time = False
                else:
                    data = pd.concat((data, ret), ignore_index=True)

    time_mean = data["e_time"].mean()
    print(f"{alg} : {time_mean.round(4)} s")
```

```python
seeds = range(5)
len_alg1 = 0
len_l1psd = 0

for seed in seeds:
    file_path = f"../results/recourse/lr_sba_Alg1_0.1_0.1_{seed}.pkl"
    ret_alg1 = pd.read_pickle(file_path)
    len_alg1 += len(ret_alg1)

    file_path = f"../results/recourse/lr_sba_L1PSD_0.1_0.1_{seed}.pkl"
    ret_l1psd = pd.read_pickle(file_path)
    len_l1psd += len(ret_l1psd)

print(f"Calculated Instances: {len_l1psd}")
print(f"Total Instances: {len_alg1}")
print(f"Percentage: {len_l1psd / len_alg1 :.2f}")
```
